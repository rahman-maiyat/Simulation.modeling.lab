```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

Assignment 1: Basic Matrix Operations

Create two 3×3 matrices with random integers in Python.

Perform addition, subtraction, and multiplication operations on these matrices.

Compute the determinant, inverse, and rank of one of the matrices.

```python
A = np.random.randint(1, 11, size=(3, 3))
B = np.random.randint(1, 11, size=(3, 3))

print(" Matrix A:")
print(A)
print(" Matrix B:")
print(B)


add_res = A + B
sub_res = A - B
mult_res = np.dot(A, B)


print("Addition:\n", add_res)
print("Subtraction:\n", sub_res)
print("Multiplication (Dot Product):\n", mult_res)
print("Matrix A ")


det_A = np.linalg.det(A)
print(f"Determinant: {det_A:.2f}")


rank_A = np.linalg.matrix_rank(A)
print(f"Rank: {rank_A}")

if det_A != 0:
    inv_A = np.linalg.inv(A)
    print("Inverse:\n", inv_A)
else:
    print("Inverse: Cannot compute (Matrix is singular)")
```

Assignment 2: Matrix Plotting and Visualization

Generate and Plot Two Random Vectors:

a. Create two vectors, each containing 15 random floats of your choice.

b. Plot these vectors on the same graph using matplotlib to compare how they look.

c. Label the axes appropriately and give the plot a title.

```python
v1 = np.random.uniform(1, 11, 15)
v2 = np.random.uniform(0, 11, 15)

#  Setup the Plot
plt.figure(figsize=(10, 5))

# Plot V1 with circles and a solid line
plt.plot(v1, marker='o', linestyle='-', color='royalblue', label="Vector 1")

# Plot V2 with squares and a dashed line
plt.plot(v2, marker='s', linestyle='--', color='darkorange', label="Vector 2")

#  Labeling and Aesthetics
plt.title('Comparison of Two Random Vectors', fontsize=14)
plt.xlabel('Index (Position in Vector)', fontsize=12)
plt.ylabel('Random Value', fontsize=12)

# Add a grid and a legend so we know which line is which
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

#  Show the final result
plt.show()
```

Assignment 2:

Generate and Visualize a Random Matrix:

a. Create a 4x4 matrix with random values of your choice.

b. Display the matrix using a heatmap with appropriate color mapping (e.g., seaborn or matplotlib).

c. Add labels to the heatmap to indicate row and column numbers.

```python
matrix = np.random.rand(4, 4)
print("Generated 4x4 Matrix:", matrix)


plt.figure(figsize=(8, 6))


sns.heatmap(matrix,
            annot=True,
            fmt=".2f",
            cmap='coolwarm',
            linewidths=0.5,
            linecolor='black',
            xticklabels=['Col 1', 'Col 2', 'Col 3', 'Col 4'],
            yticklabels=['Row 1', 'Row 2', 'Row 3', 'Row 4'])

# Add a title to explain the graph
plt.title('Heatmap Visualization of a 4x4 Matrix', fontsize=14, pad=15)

# Show the final plot
plt.show()
```

Assignment 2:

Matrix Operations and Visualization:

a. Create two 4x4 matrices with random values of your choice.

b. Perform the following operations:

i. Matrix Addition

ii. Matrix Subtraction

iii. Matrix Multiplication (ensure matrices are compatible for multiplication)

c. For each of these operations, visualize the result using bar plots. Each bar should represent one element of the matrix.

```python
A = np.random.randint(0, 11, size=(4, 4))
B = np.random.randint(0, 11, size=(4, 4))

print("Matrix A:", A)
print("Matrix B:", B)

# Matrix operations
matrix_sum = A + B
matrix_diff = A - B
matrix_prod = np.dot(A, B)

# Bar plot to visualize results
fig, axes = plt.subplots(1, 3, figsize=(18, 5))


indices = range(1, 17)

# 1. Visualize Addition (SkyBlue)
axes[0].bar(indices, matrix_sum.flatten(), color='skyblue', edgecolor='white')
axes[0].set_title('Matrix Addition (A + B)')
axes[0].set_xticks(indices)

# 2. Visualize Subtraction (salmon)
axes[1].bar(indices, matrix_diff.flatten(), color='salmon', edgecolor='white')
axes[1].set_title('Matrix Subtraction (A - B)')
axes[1].set_xticks(indices)

# 3. Visualize Multiplication (Purple)
axes[2].bar(indices, matrix_prod.flatten(), color='mediumpurple', edgecolor='white')
axes[2].set_title('Matrix Multiplication (A @ B)')
axes[2].set_xticks(indices)

# Add common labels and clean up the layout
for ax in axes:
    ax.set_xlabel('Element Index (1-16)')
    ax.set_ylabel('Value')
    ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout() # Prevents labels from overlapping
plt.show()
```
