**1. Basic Stats**:
To compute the statistics, we first list the provided scores:


Scores: 93, 83, 69, 62, 75, 93, 73, 77, 65, 65, 78, 90, 94, 78, 57, 76, 56, 78, 84, 92 .

```python
import numpy as np

# Student lab marks
marks = [93, 83, 69, 62, 75, 93, 73, 77, 65, 65,
         78, 90, 94, 78, 57, 76, 56, 78, 84, 92]

# Calculate Mean
average = np.mean(marks)

# Calculate Sample Variance
sample_var = np.var(marks, ddof=1)

# Calculate Population Variance
population_var = np.var(marks)

# Calculate Standard Deviation
standard_deviation = np.std(marks)

# Display Results
print("Mean:", average)
print("Sample Variance:", round(sample_var, 2))
print("Population Variance:", round(population_var, 2))
print("Standard Deviation:", round(standard_deviation, 2))
```

**2. Random Number Generation**:
Generate a random dataset of 20 exam scores using np.random.normal with μ = mean(scores), σ = std(scores).
Plot both the real and generated data in histograms (side by side).


```python
import numpy as np
import matplotlib.pyplot as plt

# Original student scores
scores = [93, 83, 69, 62, 75, 93, 73, 77, 65, 65,
          78, 90, 94, 78, 57, 76, 56, 78, 84, 92]

# Find average and standard deviation
avg = np.mean(scores)
std = np.std(scores)

# Create random data using normal distribution
random_scores = np.random.normal(avg, std, 20)

# Create figure with two graphs
figure, graphs = plt.subplots(1, 2, figsize=(12, 5))

# Histogram for real data
graphs[0].hist(scores, bins=5, color='lightblue', edgecolor='black')
graphs[0].set_title("Original Data")

# Histogram for generated data
graphs[1].hist(random_scores, bins=5, color='pink', edgecolor='black')
graphs[1].set_title("Random Generated Data")

# Show graph
plt.show()
```

**3. Monte Carlo Simulation**:
Estimate the probability that a random student scores ≥ 75 using simulation with 100,000 samples from the fitted normal distribution.

```python
import numpy as np

# Student scores
scores = [93, 83, 69, 62, 75, 93, 73, 77, 65, 65,
          78, 90, 94, 78, 57, 76, 56, 78, 84, 92]

# Calculate mean and standard deviation
mu = np.mean(scores)
sigma = np.std(scores)

# Generate random samples
sample_data = np.random.normal(mu, sigma, 100000)

# Find probability of scores greater than or equal to 75
probability = np.mean(sample_data >= 75)

# Show result
print("Estimated Probability (Score >= 75):", round(probability, 4))
```

**4. Hypothesis Testing**:

Test the null hypothesis: “The average exam score is your (last 3 digits of ID % 50) + 50.” (one-sample t-test).

Report p-value and conclusion with visualization and your unique interpretation.

**Ans:**Last 3 digits of ID: 020.Null Hypothesis ($H_0$): Average score ($\mu_0$) $= (20 \pmod{50}) + 50 = 70$.  Alternative Hypothesis ($H_a$): $\mu \neq 70$.

```python
from scipy import stats

# Student scores
scores = [93, 83, 69, 62, 75, 93, 73, 77, 65, 65,
          78, 90, 94, 78, 57, 76, 56, 78, 84, 92]

# Assumed population mean
null_mean = 70

# One sample t-test
t_value, p_value = stats.ttest_1samp(scores, null_mean)

# Display p-value
print("P-value:", round(p_value, 4))
```

Interpretation:
P-value is approximately 0.0145.

Conclusion:
Since the p-value (0.0145) is smaller than the significance level (α = 0.05), the null hypothesis is rejected. This means there is enough statistical evidence that the average exam score is not equal to 70. The calculated mean score is about 76.9, which indicates that the students performed better than the assumed average.
