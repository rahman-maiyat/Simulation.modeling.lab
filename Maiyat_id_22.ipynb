Basic Stats
Compute mean,variance and standard deviation .

Compare population variance** (÷ N) vs sample variance (÷ N−1).

```python
# Daily customer arrivals
arrivals = np.array([8, 12, 10, 15, 9, 13, 11, 14, 7, 16,
                     10, 12, 9, 15, 11, 13, 14, 8, 12, 10])

# Mean
average = np.mean(arrivals)

# Population calculations
population_var = np.var(arrivals)
population_std = np.std(arrivals)

# Sample calculations
sample_var = np.var(arrivals, ddof=1)
sample_std = np.std(arrivals, ddof=1)

total = len(arrivals)

print("=" * 50)
print("     DAILY CUSTOMER ARRIVAL STATISTICS")
print("=" * 50)

print(f"Mean                    : {average:.2f}")
print(f"Population Variance     : {population_var:.2f}")
print(f"Sample Variance         : {sample_var:.2f}")
print(f"Population Std Deviation: {population_std:.2f}")
print(f"Sample Std Deviation    : {sample_std:.2f}")

print("=" * 50)

print("\nExplanation:")
print("Population variance uses total data (N).")
print("Sample variance uses (N-1).")
print("Sample variance is slightly higher because")
print("it reduces estimation error in sample data.")
```

 Monte Carlo Simulation
Estimate **P(daily arrivals > 15)** using **100,000 samples** from the fitted uniform distribution.

```python
import numpy as np

# Minimum and maximum values
low = 7
high = 16

# Number of simulations
total_simulations = 100000

# Random seed
np.random.seed(0)

# Generate random samples using uniform distribution
samples = np.random.uniform(low,
                            high + 1,
                            total_simulations)

# Monte Carlo probability
mc_probability = np.mean(samples > 15)

# Exact probability calculation
greater_values = [x for x in range(low, high + 1) if x > 15]

exact_probability = len(greater_values) / (high - low + 1)

print("=" * 50)
print("      MONTE CARLO SIMULATION")
print("=" * 50)

print("Total Simulations :", total_simulations)
print("Uniform Range     :", low, "to", high)
print("Condition          : arrivals > 15")
print("Values Above 15    :", greater_values)

print("\nMonte Carlo Probability :",
      round(mc_probability, 5))

print("Exact Probability       :",
      round(exact_probability, 5))

print("Difference              :",
      round(abs(mc_probability - exact_probability), 5))

print("=" * 50)

print("\nInterpretation:")
print("The probability of getting more than")
print("15 customers is very low.")
print("Monte Carlo result is very close")
print("to the exact probability value.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# -----------------------------
# Setup data (so code runs safely)
# -----------------------------

arrivals = np.array([8, 12, 10, 15, 9, 13, 11, 14, 7, 16,
                     10, 12, 9, 15, 11, 13, 14, 8, 12, 10])

low = int(arrivals.min())
high = int(arrivals.max())

# Monte Carlo simulation
np.random.seed(0)
mc_samples = np.random.uniform(low, high + 1, 100000)

prob_mc = np.mean(mc_samples > 15)

# Exact probability
values_above = [v for v in range(low, high + 1) if v > 15]
prob_exact = len(values_above) / (high - low + 1)

# -----------------------------
# Plot
# -----------------------------

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
fig.suptitle('Monte Carlo Simulation — P(Daily Arrivals > 15)',
             fontsize=14, fontweight='bold')

# LEFT: Monte Carlo histogram
axes[0].hist(mc_samples, bins=200, color='#378ADD',
             edgecolor='none', alpha=0.7, density=True)

axes[0].axvline(15, color='#E24B4A', linestyle='--', linewidth=2)

axes[0].set_title(f'100,000 MC Samples\nUniform({low}, {high})')
axes[0].set_xlabel('Daily Arrivals')
axes[0].set_ylabel('Density')

blue_patch = mpatches.Patch(color='#378ADD',
                             label=f'P(≤15) = {1 - prob_mc:.4f}')
red_patch = mpatches.Patch(color='#E24B4A',
                           label=f'P(>15) = {prob_mc:.4f}')

axes[0].legend(handles=[blue_patch, red_patch])
axes[0].grid(alpha=0.3)

# RIGHT: Discrete PMF
x_vals = np.arange(low, high + 1)
probs = np.ones(len(x_vals)) / len(x_vals)

colors = ['#E24B4A' if v > 15 else '#378ADD' for v in x_vals]

axes[1].bar(x_vals, probs, color=colors, edgecolor='white')

axes[1].axvline(15.5, color='#E24B4A', linestyle='--', linewidth=2)

axes[1].set_title('Discrete Uniform PMF')
axes[1].set_xlabel('Daily Arrivals')
axes[1].set_ylabel('Probability')
axes[1].set_xticks(x_vals)

blue2 = mpatches.Patch(color='#378ADD',
                       label=f'P(≤15) = {1 - prob_exact:.2f}')
red2 = mpatches.Patch(color='#E24B4A',
                      label=f'P(>15) = {prob_exact:.2f}')

axes[1].legend(handles=[blue2, red2])
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

Hypothesis Testing (One-Sample t-Test)

Student ID: 022

Hypothesized Mean:
(022 mod 10) + 10 = (2) + 10 = 12

H₀: μ = 12  (mean daily arrivals equals 12)
H₁: μ ≠ 12  (two-tailed test, α = 0.05)

```python
# H0 calculation for Student ID 022

STUDENT_ID = 22   # 022

last_3_digits = STUDENT_ID % 1000

null_mean = (last_3_digits % 10) + 10

print("Student ID              : 022")
print(f"Last 3 digits           : {last_3_digits:03d}")
print(f"Last 3 digits mod 10    : {last_3_digits % 10}")
print(f"Hypothesized Mean (μ₀)  : {last_3_digits % 10} + 10 = {null_mean}")

print("\nH₀ (Null Hypothesis)    : μ = {0}".format(null_mean))
print("H₁ (Alternative)        : μ ≠ {0} (two-tailed, α = 0.05)".format(null_mean))
```

```python
import numpy as np
from scipy import stats

# -------------------------
# Data
# -------------------------
arrivals = np.array([8, 12, 10, 15, 9, 13, 11, 14, 7, 16,
                     10, 12, 9, 15, 11, 13, 14, 8, 12, 10])

# -------------------------
# Student ID = 022
# -------------------------
STUDENT_ID = 22

last_3_digits = STUDENT_ID % 1000
null_mean = (last_3_digits % 10) + 10   # = 12

# -------------------------
# Basic stats (needed for CI)
# -------------------------
mean = np.mean(arrivals)
samp_std = np.std(arrivals, ddof=1)

# -------------------------
# t-test
# -------------------------
t_stat, p_value = stats.ttest_1samp(arrivals, null_mean)

alpha = 0.05
df = len(arrivals) - 1
t_critical = stats.t.ppf(1 - alpha/2, df=df)

# Confidence interval
se = samp_std / np.sqrt(len(arrivals))
ci_low = mean - t_critical * se
ci_high = mean + t_critical * se

inside = ci_low <= null_mean <= ci_high

# -------------------------
# Output
# -------------------------
print("=" * 55)
print("ONE-SAMPLE t-TEST RESULTS (Student ID: 022)")
print("=" * 55)

print("Sample Mean (x̄)      :", round(mean, 4))
print("Null Mean (μ₀)        :", null_mean)
print("t-statistic           :", round(t_stat, 4))
print("p-value               :", round(p_value, 6))
print("t-critical            :", round(t_critical, 4))
print("95% CI                :", [round(ci_low, 4), round(ci_high, 4)])
print("μ₀ inside CI?         :", inside)

print("\nDecision:")

if p_value < alpha:
    print("REJECT H₀")
else:
    print("FAIL TO REJECT H₀")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# -------------------------
# Data
# -------------------------
arrivals = np.array([8, 12, 10, 15, 9, 13, 11, 14, 7, 16,
                     10, 12, 9, 15, 11, 13, 14, 8, 12, 10])

# -------------------------
# Student ID = 022
# -------------------------
STUDENT_ID = 22
last_3_digits = STUDENT_ID % 1000
null_mean = (last_3_digits % 10) + 10   # = 12

# -------------------------
# Stats needed
# -------------------------
mean = np.mean(arrivals)
samp_std = np.std(arrivals, ddof=1)
se = samp_std / np.sqrt(len(arrivals))

df = len(arrivals) - 1
alpha = 0.05
t_stat, p_value = stats.ttest_1samp(arrivals, null_mean)
t_critical = stats.t.ppf(1 - alpha/2, df=df)

ci_low = mean - t_critical * se
ci_high = mean + t_critical * se

# -------------------------
# Plot
# -------------------------
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

fig.suptitle(
    f'Hypothesis Test | H₀: μ = {null_mean} vs H₁: μ ≠ {null_mean} (Student ID: 022)',
    fontsize=14, fontweight='bold'
)

# ─────────────────────────────
# Plot 1: t distribution
# ─────────────────────────────
ax1 = axes[0, 0]

x = np.linspace(-6, 6, 600)
y = stats.t.pdf(x, df)

ax1.plot(x, y, color='#378ADD', linewidth=2.5)

ax1.fill_between(x[x >= t_critical],
                 stats.t.pdf(x[x >= t_critical], df),
                 color='#E24B4A', alpha=0.6)

ax1.fill_between(x[x <= -t_critical],
                 stats.t.pdf(x[x <= -t_critical], df),
                 color='#E24B4A', alpha=0.6)

ax1.axvline(t_stat, color='#533AB7', linewidth=2, label=f't = {t_stat:.3f}')
ax1.axvline(t_critical, color='#E24B4A', linestyle='--')
ax1.axvline(-t_critical, color='#E24B4A', linestyle='--')

ax1.set_title("t-Distribution")
ax1.legend()
ax1.grid(alpha=0.3)

# ─────────────────────────────
# Plot 2: Mean comparison
# ─────────────────────────────
ax2 = axes[0, 1]

ax2.bar(['Sample Mean', 'Hypothesized Mean'],
        [mean, null_mean],
        color=['#378ADD', '#E24B4A'])

ax2.set_title("Mean Comparison")
ax2.grid(axis='y', alpha=0.3)

# ─────────────────────────────
# Plot 3: Confidence Interval
# ─────────────────────────────
ax3 = axes[1, 0]

x_ci = np.linspace(mean - 5*se, mean + 5*se, 500)
y_ci = stats.norm.pdf(x_ci, mean, se)

ax3.plot(x_ci, y_ci, color='#378ADD')

ax3.fill_between(x_ci[(x_ci >= ci_low) & (x_ci <= ci_high)],
                 stats.norm.pdf(x_ci[(x_ci >= ci_low) & (x_ci <= ci_high)], mean, se),
                 color='#378ADD', alpha=0.3)

ax3.axvline(mean, color='#378ADD', linestyle='--')
ax3.axvline(null_mean, color='#E24B4A', linestyle='--')

ax3.set_title("95% Confidence Interval")
ax3.grid(alpha=0.3)

# ─────────────────────────────
# Plot 4: Raw data
# ─────────────────────────────
ax4 = axes[1, 1]

ax4.plot(range(1, 21), arrivals, 'o-', color='#378ADD')

ax4.axhline(mean, color='#378ADD', linestyle='--')
ax4.axhline(null_mean, color='#E24B4A', linestyle='--')

ax4.set_title("Daily Arrivals")
ax4.set_xticks(range(1, 21))
ax4.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
