2. Simulation

○ Count the fraction of points that fall inside the ellipse.

○ Estimate area using:

● Rectangle area = (2 \times 1 = 2)

```python
import random

def estimate_area(N):
    inside = 0
    a = 2
    b = 1

    for _ in range(N):
        input = random.uniform(0,2)
        output = random.uniform(0,1)

        if (input*input)/(a*a) + (output*output)/(b*b) <= 1:
            inside += 1

    rectangle_area = 2
    area = (inside/N) * rectangle_area
    return area
```

3. Experiment

○ Use N = 100, 1000, 10000 points

○ Compare your estimate with the true area of the quarter ellipse:

```python
print("Estimated area for 100 points is:", estimate_area(100))
print("Estimated area for 1000 points is:", estimate_area(1000))
print("Estimated area for 10000 points is:", estimate_area(10000))

```

4. Analysis

○ How does the estimate improve with larger N?

○ Why does Monte Carlo work for this problem?

As the number of random points increases, the estimated area becomes closer to the true value. This happens because more samples give a better approximation. Monte Carlo works by using probability and random sampling to estimate the area of geometric shapes.
